## sampling and diff types distribution

```python
a=np.array([1,2,3,4,8,6,4,6,7])
#random sampling
print("random sampling",np.random.choice(a,4))
#uniform distribution
print("uniform distribution",np.random.uniform(1,2,2))
print("uniform distribution",np.random.uniform(1,1000,6))

#normal distribution
print("normal distribution",np.random.normal(1,10,5))
#binomal distribution
#there are trail =7,and success prob=50% ,size =7
print("binomal distribution",np.random.binomial(7,0.5,7))
```

```python
#shuffle and pemutation
arr=np.arange(10)
print(arr)
np.random.shuffle(arr)
print(arr)
print(np.random.permutation(arr))
```

```python
#excerise train test split
arr=np.arange(1,21)
print(arr)
train=np.random.choice(arr,12,replace=False)
test=np.setdiff1d(arr,train)
print(train)
print(test)
```
