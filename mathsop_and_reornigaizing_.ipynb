## linear algebra


### Essential Linear Algebra Operations
Using NumPy's `linalg` submodule, we can perform advanced matrix math. This section demonstrates:
* **Dot Product (`np.dot`):** Multiplying matrices (often utilizing transposes `.T` to align dimensions).
* **Matrix Determinant (`np.linalg.det`):** Computing the scaling factor of the transformation.
* **Matrix Inverse (`np.linalg.inv`):** Finding the matrix that yields the identity matrix when multiplied by the original.

```python
a=np.array([[1,2,3],[4,5,6]])
b=np.array([[1,3,5],[7,9,11]])
print("multiplication :",np.dot(a,b.T))
print("addition",np.add(a,b))
print("transpose",a.T,"\n",b.T)

```

```python
d=np.array([[2,4],[6,8]])
z=np.array([[1,2,],[3,4]])
print("determinant",np.linalg.det(d))
print("inverse",np.linalg.inv(d))
```

## statistics basic

### Statistical Reductions and Axis Operations
This section highlights how to compute summary statistics on our datasets:
* **`np.min()` and `np.max()`**: Finding boundary values overall or along specific dimensions.
* **`np.sum()`**: Accumulating all values in an array.
* **Understanding Axes**: Operating along `axis=0` (vertically down columns) versus `axis=1` (horizontally across rows).

```python
stat=np.array([[1,2,3],[4,5,6]])
print(stat)
print(np.min(stat,axis=0))
print(np.max(stat,axis=0))
print(np.sum(stat))
```

```python
# Calculate the min and max along axis 1 (across columns)
print("Min along axis 1:", np.min(stat, axis=0))
print("Max along axis 1:", np.max(stat, axis=0))
```

### Reorganizing, Reshaping, and Stacking Arrays
Often, data needs to be restructured before processing. Here we cover:
* **`reshape()`**: Changing the dimensions of an array without changing its data.
* **`np.vstack()`**: Stacking arrays sequentially on top of each other (vertically).
* **`np.hstack()`**: Combining arrays end-to-end (horizontally).

```python
beforearr=np.array([[2,4,6,8],[3,7,9,11]])
after=beforearr.reshape(2,2,2)
print(after)
```

```python
#vertical stack
v=np.array([1,2,3,5])
v2=np.array([9,8,7,6])
np.vstack(v)
np.vstack(v2)
#np.vstack([v,v2,v,v2])
```

```python
#horizanotal stack
h1b=np.array([2,4,6,8])
h2b=np.array([1,3,5,7])
np.hstack([h1b,h2b])
```
