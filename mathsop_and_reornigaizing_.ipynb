# Mathematical operation by using numpy

```python
import numpy as np
print("Mathematical operation")
```

```python
a=np.array([1,2,3,4,5,6,7,8])
print(a)
```

```python
b=np.array([11,22,33,44,55,66,77,88])
c=np.array([])
c=a+b
print(c)
```

```python
z=np.array([[12,3,3,4],[23,4,56,78]])
print(z[z>10])
```

```python
a=np.ones((2,2))
b=np.full((2,3),2)
print(a)
print(b)
c=np.matmul(a,b)
print(c)
np.linalg.det(b)
```

```python
c=np.identity(4)
print(c)
np.linalg.det(c)
```

## linear algebra


```python
a=np.array([[1,2,3],[4,5,6]])
b=np.array([[1,3,5],[7,9,11]])
print("multiplication :",np.dot(a,b.T))
print("addition",np.add(a,b))
print("transpose",a.T,"\n",b.T)

```

```python
d=np.array([[2,4],[6,8]])
z=np.array([[1,2,],[3,4]])
print("determinant",np.linalg.det(d))
print("inverse",np.linalg.inv(d))
```

## statistics basic

```python
stat=np.array([[1,2,3],[4,5,6]])
print(stat)
print(np.min(stat,axis=0))
print(np.max(stat,axis=0))
print(np.sum(stat))
```

```python
# Calculate the min and max along axis 1 (across columns)
print("Min along axis 1:", np.min(stat, axis=0))
print("Max along axis 1:", np.max(stat, axis=0))
```
